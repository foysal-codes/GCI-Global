```python
URL="D:\Downloads\GHCND_sample_csv.csv"
```

```python
raw=np.loadtxt(URL,delimiter=",",usecols=[6,7,8],skiprows=1)
```

```python
week_one_tmax = raw[0:7,0]
week_one_tmin = raw[0:7,1]
week_one_prcp = raw[0:7,2]
week_two_tmax = raw[7:14,0]
print("Week 1 TMAX:",week_one_tmax)
print("Week 1 TMIN:",week_one_tmin)
print("Week 1 PRCP: ",week_one_prcp)
```

```python
print("Size of week 1 TMAX:",len(week_one_tmax))
print("Size of week 1 TMAX:",week_one_tmax.shape)
```

```python
a = np.array([1,2,3,4,5])
b= np.array([2,2,3,6,1])
print("Sum:",a+b)
print("Difference:",a-b)
print("Product:",a*b)
print("Quotient:",a/b)
```

```python
week_one_tmax /=10
week_one_tmin /= 10
week_one_prcp /= 10
print("Week 1 TMAX (in celcius):",week_one_tmax)
print("Week 1 TMIN (in celcius):",week_one_tmin)
print("Week 1 PRCP (in mm):",week_one_prcp)
```

```python
print("Hottest day in week 1:",np.max(week_one_tmax),"Degree Celcius")
print("Total rainfall in week 1: ",np.sum(week_one_prcp),"mm")
print("Average TMAX in week 1: ",np.mean(week_one_tmax),"Degree Celcius")
print("Standard Deviation of TMAX in week 1: ",np.std(week_one_tmax))
```
